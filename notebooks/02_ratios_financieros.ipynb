Objetivo: calcular márgenes, rentabilidad y posición de caja, y **comentar qué significan**.
Lo importante no es el número, sino tu interpretación.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))  # para poder importar src/

import pandas as pd
import matplotlib.pyplot as plt
from src.ratios import calcular_ratios

raw = Path("../data/raw")
res = pd.read_csv(raw / "resultados.csv", index_col=0)
bal = pd.read_csv(raw / "balance.csv", index_col=0)
flu = pd.read_csv(raw / "flujos.csv", index_col=0)
```

```python
ratios = calcular_ratios(res, bal, flu)
ratios.to_csv("../data/processed/ratios.csv")

pct = [c for c in ratios.columns if "(M€)" not in c]
ratios.style.format({c: "{:.1%}" for c in pct} | {c: "{:,.0f}" for c in ratios.columns if c not in pct})
```

## 1. Márgenes

```python
ax = ratios[["Margen bruto", "Margen EBITDA", "Margen EBIT", "Margen neto"]].plot(
    marker="o", figsize=(9, 4), title="Inditex · Márgenes")
ax.yaxis.set_major_formatter(lambda x, _: f"{x:.0%}"); ax.grid(alpha=.3)
plt.savefig("../img/margenes.png", dpi=150, bbox_inches="tight"); plt.show()
```

## 2. Rentabilidad

```python
ax = ratios[["ROE", "ROA"]].plot(kind="bar", figsize=(9, 4), title="Inditex · Rentabilidad")
ax.yaxis.set_major_formatter(lambda x, _: f"{x:.0%}"); ax.grid(alpha=.3, axis="y")
plt.savefig("../img/rentabilidad.png", dpi=150, bbox_inches="tight"); plt.show()
```
