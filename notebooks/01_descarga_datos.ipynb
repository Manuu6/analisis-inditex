# 01 · Descarga de datos de Inditex

Objetivo: descargar precios históricos y estados financieros de Inditex (`ITX.MC`) y guardarlos en `data/raw/`.

> **Nota:** el ejercicio fiscal de Inditex termina el **31 de enero**. Un cierre "2025-01-31" corresponde al ejercicio 2024.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))  # para poder importar src/

from src.datos import TICKER, descargar_precios, descargar_estados, guardar
```

## 1. Precios históricos

```python
precios = descargar_precios(TICKER, inicio="2015-01-01")
precios.tail()
```

```python
import matplotlib.pyplot as plt

ax = precios["Close"].plot(figsize=(10, 4), title="Inditex · Precio ajustado (€)")
ax.set_xlabel(""); ax.grid(alpha=.3)
plt.savefig("../img/precio_historico.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 2. Estados financieros

```python
estados = descargar_estados(TICKER)
for nombre, df in estados.items():
    print(nombre, df.shape)
estados["resultados"].head(15)
```

## 3. Guardar en `data/raw/`

```python
guardar(precios, "precios")
for nombre, df in estados.items():
    print("Guardado:", guardar(df, nombre))
```
